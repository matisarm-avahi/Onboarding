Exercise 1: First call to Claude

```python
import json

bedrock_runtime = boto3.client('bedrock-runtime', region_name='us-east-1')

def invoke_claude(prompt, max_tokens=500, model_id='us.anthropic.claude-sonnet-4-5-20250929-v1:0'):
    body = json.dumps({
        "anthropic_version": "bedrock-2023-05-31",
        "max_tokens": max_tokens,
        "messages": [
            {
                "role": "user",
                "content": prompt
            }
        ]
    })
    start = time.time()
    response = bedrock_runtime.invoke_model(
        modelId=model_id,
        body=body
    )

    response_body = json.loads(response['body'].read())
    end = time.time()
    print(f"Response time: {end - start:.2f} seconds")
    return response_body['content'][0]['text']

# Example usage
result = invoke_claude("Explain what AWS Bedrock is in 3 paragraphs")
print(result)
```

```python
# Example usage
result = invoke_claude("How can I build a mini drone using raspberry pi and arduino?", max_tokens=1000)
print(result)
```

```python
# Example usage
result = invoke_claude("What does an AI Engineer do?", max_tokens=100)
print(result)
```

Exercise 2: Model comparison

```python
# Example usage
sonnet = invoke_claude("Give me an overview of the Mexican culture in 3 paragraphs", max_tokens=150) #Sonnet, best suitable for general-purpose content
haiku = invoke_claude("Give me an overview of the Mexican culture in 3 paragraphs", max_tokens=150, model_id='us.anthropic.claude-haiku-4-5-20251001-v1:0') #Haiku, best suitable for short-form content
opus = invoke_claude("Give me an overview of the Mexican culture in 3 paragraphs", max_tokens=150, model_id='us.anthropic.claude-opus-4-6-v1') #Opus, best suitable for long-form content

print("Sonnet Model Output:\n", sonnet)
print("\nHaiku Model Output:\n", haiku)
print("\nOpus Model Output:\n", opus)
```

Exercise 3: Parameter handling

```python
def invoke_claude_temp(prompt, max_tokens=500, temperature=0.7, top_k=15, model_id='us.anthropic.claude-sonnet-4-5-20250929-v1:0'):
    body = json.dumps({
        "anthropic_version": "bedrock-2023-05-31",
        "max_tokens": max_tokens,
        "temperature": temperature,
        "top_k": top_k,
        "messages": [
            {
                "role": "user",
                "content": prompt
            }
        ]
    })
    start = time.time()
    response = bedrock_runtime.invoke_model(
        modelId=model_id,
        body=body
    )

    response_body = json.loads(response['body'].read())
    end = time.time()
    print(f"Response time: {end - start:.2f} seconds")
    return response_body['content'][0]['text']

def invoke_claude_toppk(prompt, max_tokens=500, top_k=15, top_p=0.9, model_id='us.anthropic.claude-sonnet-4-5-20250929-v1:0'):
    body = json.dumps({
        "anthropic_version": "bedrock-2023-05-31",
        "max_tokens": max_tokens,
        "top_k": top_k,
        "top_p": top_p,
        "messages": [
            {
                "role": "user",
                "content": prompt
            }
        ]
    })
    start = time.time()
    response = bedrock_runtime.invoke_model(
        modelId=model_id,
        body=body
    )

    response_body = json.loads(response['body'].read())
    end = time.time()
    print(f"Response time: {end - start:.2f} seconds")
    return response_body['content'][0]['text']

# Example usage
temp_result = invoke_claude_temp("Explain what AWS Bedrock is in 3 paragraphs")
print(temp_result)
topp_result = invoke_claude_toppk("Explain what AWS Bedrock is in 3 paragraphs")
print(topp_result)
```
